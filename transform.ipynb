```python
import json
import pandas as pd
import isodate
import json
```

```python
with open("data/YT_data_2026-05-11.json", "r", encoding="utf-8") as f:
    data = json.load(f)
```

```python
print(type(data))
```

```python
df = pd.DataFrame(data)
```

```python
df
```

```python
df["views"] = df["viewCount"].astype(int)
df["likes"] = df["likeCount"].astype(int)
df["comments"] = df["commentCount"].astype(int)
```

```python
df["duration_sec"] = df["duration"].apply(
    lambda x: isodate.parse_duration(x).total_seconds()
)
```

```python
df["published_at"] = pd.to_datetime(df["publishedAt"])
```

```python
df_final = df[[
    "videoId",
    "title",
    "published_at",
    "duration_sec",
    "views",
    "likes",
    "comments"
]]
```

```python
df_final
```

```python
df_final.to_csv("data/youtube_clean.csv", index=False)
```
